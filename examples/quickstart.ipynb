# Iragi quickstart
Install the package with `pip install iragi`, then inspect a small DataFrame.

```python
import pandas as pd
from iragi.scan import scan_dataframe

sales = pd.DataFrame({
    'order_id': [101, 102, 102, 104],
    'amount': [25.0, None, None, 80.0],
})
result = scan_dataframe(sales, critical_columns=['order_id'])
print(result.quality_score.total_score)
print(result.recommendations)
```

## Stream a CSV
Use `scan(..., stream=True)` for chunked CSV and JSON Lines scanning. Streaming scores are estimates; duplicate fingerprints are retained in memory.

```python
from pathlib import Path
import iragi

Path('sales.csv').write_text('id,amount\n1,25\n2,80\n2,80\n')
streamed = iragi.scan('sales.csv', stream=True, chunksize=2)
print(streamed.row_count, streamed.duplicate_rows_count, streamed.score_is_estimated)
```
